```python
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Datasets/flight_details.csv")
```

```python
data.head()
```

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

data['Months'] = pd.Categorical(
    data['Months'],
    categories=month_order,
    ordered=True
)
```

```python
flight_data = data.pivot(
index='Months',
columns='Years',
values='Passengers'
)
```

```python
# Plot heatmap
plt.figure(figsize=(10, 6), dpi=150)
 
sns.heatmap(
flight_data,
cmap='mako_r'
)
 
plt.title('Flight Passengers from 1949 to 1960')
plt.xlabel('Years')
plt.ylabel('Months')
 
plt.tight_layout()
plt.show()
```
